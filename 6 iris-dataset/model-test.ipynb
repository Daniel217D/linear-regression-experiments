```python
# Импорт зависимостей
import joblib
```

```python
# Импорт функций необходимых для модели
from features import make_areas, areas_feature_names_out

# Импорт модели и метаданных из файла
artifact_loaded = joblib.load('model.joblib')

model = artifact_loaded['model']
metadata = artifact_loaded['metadata']

metadata
```

```python
# Импорт датасета
from sklearn.datasets import load_iris

iris = load_iris(as_frame=True)

df = iris.data
```

```python
# Построим матрицу ошибок для тестовой выборки
from sklearn.metrics import ConfusionMatrixDisplay
from matplotlib import pyplot as plt

ConfusionMatrixDisplay.from_predictions(
    iris.target,
    model.predict(df),
    display_labels=iris.target_names
)

plt.show()
```
